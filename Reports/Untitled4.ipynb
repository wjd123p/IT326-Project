# [1] Problem


This project aims to analyze student data to address the challenge of understanding and predicting academic performance. Using multiple student-related attributes, the project applies classification techniques to categorize students into different academic performance groups based on their final grades. In addition, clustering methods are used to group students with similar characteristics and study habits, helping to uncover hidden patterns and common student profiles within the data. By combining these approaches, the project provides meaningful insights into student performance and supports data-driven decision making in education.


# [2] Data Mining Task

The data mining tasks applied in this project include classification and clustering. Classification techniques are used to categorize students into different academic performance groups based on their final grades and attributes. Clustering methods are employed to group students with similar characteristics, study habits, and profiles, uncovering hidden patterns and common trends within the educational dataset.


# [3] Data

The dataset used in this project comprises various student-related attributes, including demographic information, study habits, academic background, and performance metrics. These features serve as the input for applying classification and clustering algorithms to evaluate and understand student academic outcomes.


# [4] Data Preprocessing

Data preprocessing involves cleaning and preparing the raw student dataset to ensure data quality and model effectiveness. This step includes handling missing values, encoding categorical variables, normalizing numerical features, and selecting the most relevant attributes required for the classification and clustering models.
