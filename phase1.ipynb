**Class Attribute:**
The class attribute in this dataset is final_grade, which represents the student's final academic performance. It contains five classes: A, B, C, D, and F.

```python
print("Target Column (class attribute): final_grade")
print("Number of classes:", df['final_grade'].nunique())
print("Class values:", df['final_grade'].unique())

print("\nClass Distribution:")
print(df['final_grade'].value_counts().sort_index())
```
